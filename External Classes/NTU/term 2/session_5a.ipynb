```python
# Setup
from ugot import ugot

got = ugot.UGOT()
got.initialize("192.168.1.209")

got.load_models(["line_recognition", "face_recognition"])

print("connected!")
```

```python
# delete all names

for name in got.face_recognition_get_all_names():
    got.face_recognition_delete_name(name)
```

```python
# add in name

got.face_recognition_add_name("Name")
```

```python
# test recognition

faces = got.get_face_recognition_total_info()

if faces:
    for face in faces:
        print(f"Recognized {face[0]}!")
else:
    print("Can't see anybody :(")
```
