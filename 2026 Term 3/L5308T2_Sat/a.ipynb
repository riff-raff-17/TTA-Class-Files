```python
nums = [2,7,11,15]
target = 9

def two_sum(nums, target):
    for i, num1 in enumerate(nums):
        for j in range(i + 1, len(nums)):
            if num1 + nums[j] == target:
                return [i, j]

two_sum(nums, target)
```
