# Python List Fundamentals

## Part 1: Fundamentals

### 1.1 What is a list?

A list is an **ordered, mutable** (changeable) collection of items. Lists can hold
any type of data, and even a mix of different types.

```python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]
mixed = ["david", 21, True]
empty_list = []
```

### 1.2 Indexing

Each item in a list has a position, called an **index**, starting at `0`.
Negative indices count backward from the end of the list.

```python
fruits = ["apple", "banana", "cherry", "durian"]

print(fruits[0])  # first item
print(fruits[2])  # third item
print(fruits[-1])  # last item
print(fruits[-2])  # second to last item
```

### 1.3 Slicing

You can grab a sub-section of a list using `list[start:stop:step]`.
- `start` is inclusive
- `stop` is exclusive
- `step` is optional (default is 1)

```python
fruits = ["apple", "banana", "cherry", "durian", "elderberry"]

print(fruits[2:4])  # index 2 up to (not including) 4
print(fruits[:3])  # from the start up to index 3
print(fruits[2:])  # from index 2 to the end
print(fruits[::2])  # every second item
print(fruits[::-1])  # the whole list, reversed
```

### 1.4 Modifying lists

Lists are **mutable**, meaning you can change them in place.

```python
fruits = ["apple", "banana", "cherry", "durian", "elderberry"]

# Change an item by index
fruits[1] = "blueberry"
print(fruits)

# Add an item to the end
fruits.append("fig")
print(fruits)

# Insert an item at a specific position
fruits.insert(1, "kiwi")
print(fruits)

# Remove an item by value
fruits.remove("cherry")
print(fruits)

# Remove an item by index (and get its value back)
removed_item = fruits.pop(0)
print(removed_item, "->", fruits)
```

### 1.5 Useful list operations

A quick tour of common built-in functions and methods.

```python
numbers = [5, 3, 8, 1, 9, 2]

print(len(numbers))  # number of items
print(sum(numbers))  # sum of all items
print(max(numbers))  # largest item
print(min(numbers))  # smallest item
print(sorted(numbers))  # a NEW sorted list
print(numbers)

numbers.sort()  # sorts the list IN PLACE
print(numbers)

numbers.reverse()  # reverses the list IN PLACE
print(numbers)

print(3 in numbers)  # membership check
```

### 1.6 Looping through lists

You'll very often want to do something with every item in a list.

```python
fruits = ["apple", "banana", "cherry", "durian", "elderberry"]

# Basic loop
for i in range(len(fruits)):
    print(fruits[i])

print("---------------")

for fruit in fruits:
    print(fruit)

# Loop with index, using enumerate()
for index, fruit in enumerate(fruits):
    print(index, "->", fruit)
```
