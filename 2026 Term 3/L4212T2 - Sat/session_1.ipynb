## 1. Lists

A **list** is an ordered, *mutable* (changeable) collection of items. Think of it like a shopping list where you can add, remove, or reorder items.

Key properties:
- Ordered (items keep their position)
- Mutable (you can change them after creation)
- Allow duplicates
- Can hold mixed types (though usually you'll keep one type per list)

```python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers  = [1, 2, 3, 4, 5]
mixed = ["hello", 42, 3.14, True]

print(fruits)
print(numbers)
print(mixed)
```

### Indexing and slicing

Python uses **zero-based indexing** — the first item is at index `0`. Negative indices count from the end.

```python
fruits = ["apple", "banana", "cherry", "date", "etrog", "seth"]

print(fruits[0]) # first item
print(fruits[-1]) # last item
print(fruits[1:4]) # slice: index 1 up to (not including) 3
print(fruits[:2]) # from the start up to index 2
print(fruits[::-1]) # the whole list, reversed
```

```python
shopping_list = ["milk", "eggs", "ඞ"]

shopping_list.append("bread") # add to the end
shopping_list.insert(0, "coffee") # insert at a specific position
shopping_list.remove("eggs") # remove by value
last_item = shopping_list.pop() # remove and return the last item

print(last_item)
print(shopping_list)
```

## 2. Tuples

A **tuple** is just like a list, but **immutable** — once created, it can't be changed. This makes tuples great for data that shouldn't change, like coordinates or fixed records.

Key properties:
- Ordered
- Immutable (cannot be modified after creation)
- Allow duplicates
- Slightly faster and more memory-efficient than lists

```python
# Creating tuples
point = (3, 4)
rgb_color = (255, 0, 128)
single_item_tuple = (42,) # note the comma! Without it, this is just an int

print(rgb_color[1])
```

### Tuple unpacking

This is one of the most useful Python features — assigning multiple variables in one line.

```python
point = (3, 4)
x, y = point

print(f"x: {x}, y: {y}")

# Useful for swapping variables
a, b = 1, 2
a, b = b, a
print(f"a = {a}, b = {b}")

# Useful when looping over pairs
coordinates = [(0, 0), (1, 2), (3, 4)]
for x, y in coordinates:
    print()
```
